```python
import mlflow
from mlflow import MlflowClient
```

```python
client = MlflowClient(tracking_uri="sqlite:////workspaces/MLOps/mlflow.db")
```

```python
from mlflow.entities import ViewType
```

### Inspecting runs

```python
runs = client.search_runs(
    experiment_ids=1, # experiment index
    filter_string="",
    run_view_type=ViewType.ACTIVE_ONLY, # view only active runs
    max_results=5,
    order_by=["metrics.auc DESC"] # sort by metric in ascending order.
)
```

```python
for run in runs:
    print(f"run id: {run.info.run_id}, auc: {run.data.metrics['auc']:.4f}")
```

### Promote a model to Registry

### Transition a model to `staging`

```python
model_name= "epl-winner-classification"
```

```python
client.transition_model_version_stage(
    name=model_name,
    version=1,
    stage='Staging',
    archive_existing_versions=False
)
```
